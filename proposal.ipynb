## 1. Motivation & Problem

Wildlife-vehicle collisions can harm animals, endanger drivers, and reduce connections between wildlife habitats. Understanding where these incidents are concentrated may help transportation agencies decide where to place warning signs, fencing, wildlife crossings, or additional monitoring resources.

My research question is: **Which regions and wildlife species are most represented among Washington highway segments ranked as high wildlife-collision priorities, and how do recorded wildlife incidents differ across safety-priority levels?**

I hope to identify patterns in wildlife records, highway regions, and safety-priority categories that may support better transportation and habitat-connectivity planning.

## 2. Data

I will use the Washington State Department of Transportation’s [WAHCAP 2025 Wildlife-Related Safety dataset](https://data.wsdot.wa.gov/arcgis/rest/services/Shared/WAHabitatConnectivityActionPlan/FeatureServer/3). It is based mainly on large-animal carcass-removal records and wildlife-vehicle collision reports from 2019–2023.

The dataset includes highway route and milepost information, WSDOT region, traffic volume, wildlife safety-priority level, and counts for animals such as deer, elk, moose, bears, cougars, and coyotes. It also includes reported crash, injury, and fatality counts.

The data are publicly accessible through WSDOT’s ArcGIS service and are intended for planning use. WSDOT does not guarantee that the data are complete or suitable for every purpose, so I will cite WSDOT and clearly describe the dataset’s limitations.

The dataset does not contain personal information. However, it may undercount collisions because some animals are not found or reported. Large animals are also more likely to be recorded than small animals. Therefore, I will treat the data as recorded incidents rather than a complete measure of all wildlife deaths.

## 3. Unknowns & Dependencies

The ArcGIS service limits the number of records returned in one request, so I may need to download the data in multiple batches. Some fields may contain missing values, and zero values may mean either no recorded incidents or incomplete reporting.

The data combine records from 2019–2023, so they cannot support a detailed monthly or seasonal analysis. Geographic mapping may also require additional Python tools. If mapping becomes too complex, I will focus on comparisons by WSDOT region, animal type, and safety-priority category using tables and charts.
