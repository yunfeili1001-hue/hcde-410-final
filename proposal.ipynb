## 1. Motivation & Problem

Wildlife-vehicle collisions can harm animals, endanger drivers, and reduce connections between wildlife habitats. Understanding where these incidents are concentrated may help transportation agencies decide where to place warning signs, fencing, wildlife crossings, or additional monitoring resources.

My research question is: **Which regions and wildlife species are most represented among Washington highway segments ranked as high wildlife-collision priorities, and how do recorded wildlife incidents differ across safety-priority levels?**

I hope to identify patterns in wildlife records, highway regions, and safety-priority categories that may support better transportation and habitat-connectivity planning.

## 2. Data

I will use the Washington State Department of Transportation’s [WAHCAP 2025 Wildlife-Related Safety dataset](https://data.wsdot.wa.gov/arcgis/rest/services/Shared/WAHabitatConnectivityActionPlan/FeatureServer/3). It is based mainly on large-animal carcass-removal records and wildlife-vehicle collision reports from 2019–2023.

The dataset includes highway route and milepost information, WSDOT region, traffic volume, wildlife safety-priority level, and counts for animals such as deer, elk, moose, bears, cougars, and coyotes. It also includes reported crash, injury, and fatality counts.

The data are publicly accessible through WSDOT’s ArcGIS service and are intended for planning use. WSDOT does not guarantee that the data are complete or suitable for every purpose, so I will cite WSDOT and clearly describe the dataset’s limitations.

The dataset does not contain personal information. However, it may undercount collisions because some animals are not found or reported. Large animals are also more likely to be recorded than small animals. Therefore, I will treat the data as recorded incidents rather than a complete measure of all wildlife deaths.

## 5. Background and Related Work

Wildlife-vehicle collisions can harm animals, threaten driver safety, and increase habitat fragmentation caused by highways. WSDOT uses wildlife crossings, fencing, and other measures to improve habitat connectivity and reduce collision risks. WSDOT reports that wildlife crossings combined with fencing can reduce wildlife-vehicle collisions by approximately 80–90% on average.

WSDOT has collected location data for large-animal carcass removals since 1973. However, its documentation explains that these records are incomplete. Animals that are not found, reported, or removed by highway workers may not appear in the data. Small animals are also more likely to be missed. Therefore, the dataset represents recorded incidents rather than every wildlife death or collision that occurred.

The WAHCAP 2025 Wildlife-Related Safety layer is based mainly on large-bodied wildlife carcass-removal records and wildlife-vehicle collision reports from 2019–2023. It includes highway locations, WSDOT regions, traffic volume, wildlife-safety scores and priorities, and records for species such as deer, elk, moose, black bears, cougars, and coyotes.

These limitations are important when comparing regions. A region with fewer records may not necessarily have fewer actual incidents. Its results may also be affected by animal size, reporting practices, highway monitoring, or maintenance procedures.

## 6. Methodology

### 6.1 Data Collection

I will download the data through the WSDOT ArcGIS REST API. The layer contains 7,041 records, while the service returns no more than 2,000 records per request. I will download the records in four batches and combine them into one Pandas DataFrame.

The main analysis does not require geographic geometry, so I will initially download only the attribute fields. A map will be treated as an optional extension.

### 6.2 Main Variables

The analysis will use the following fields:

- `Safetyv10_Priority`: High, Medium, Low, or No Rank
- `WRSv10`: wildlife-related safety score
- `WSDOTRegion`: WSDOT region
- `RouteID` and `RouteType`: highway route and route type
- `AADT_1`: average annual daily traffic
- `DeerCount`, `ElkCount`, `MooseCount`, `BlackBearCount`, `CougarCount`, and `CoyoteCount`: recorded wildlife counts
- `DeerCrashCount` and `ElkCrashCount`: recorded deer and elk crashes
- `CrashInjCount` and `CrashFatalCount`: recorded injury and fatal crashes

The official documentation does not clearly define `LCARCount`, so I will not treat it as a total count of all large-animal records unless I can confirm its meaning.

### 6.3 Data Cleaning

I will:

1. Check for duplicate records using `OBJECTID`.
2. Confirm that the priority, region, traffic, and count fields have the correct data types.
3. Check for negative counts, unreasonable traffic values, and other possible errors.
4. Keep `No Rank` as a separate category instead of treating it as Low.
5. Recheck missing values after downloading the data. The core fields currently contain no missing values.
6. Treat zero as no recorded incident while recognizing that incomplete reporting may also produce low or zero counts.
7. Analyze wildlife records and police crash records separately to avoid combining different sources or double-counting incidents.

### 6.4 Analysis

First, I will calculate the number and percentage of highway segments in each priority category.

Next, I will create a cross-tabulation of `WSDOTRegion` and `Safetyv10_Priority`. For each region, I will calculate the percentage of its segments that are ranked High priority. This is more appropriate than comparing only raw counts because the regions contain different numbers of highway segments.

For the species analysis, I will total the species-specific fields within High-priority segments and create a ranked bar chart. These values will be described as recorded species totals in the dataset, not as complete counts of wildlife deaths in Washington.

Finally, I will compare `AADT_1`, wildlife counts, `DeerCrashCount`, and `ElkCrashCount` across priority levels. I will report both means and medians because the distributions may be skewed. I will use box plots or a logarithmic scale when appropriate. Injury and fatal crash counts will be treated as supplementary findings because they are relatively rare.
